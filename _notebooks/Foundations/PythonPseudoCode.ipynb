```python
animeTitles ← ["Naruto", "Attack on Titan", "Death Note", "One Piece"]
animeRatings ← [8, 9, 9, 10]
animeStatus ← ["Watching", "Completed", "Completed", "Watching"]

DISPLAY("Welcome to the Anime Tracker!")
DISPLAY("You are currently tracking " + LENGTH(animeTitles) + " anime.")

newTitle ← INPUT("Enter the anime title you want to add: ")
newRating ← INPUT("Enter your rating (1-10): ")
newStatus ← INPUT("Enter status (Watching/Completed/Dropped): ")

totalRating ← 0
count ← LENGTH(animeRatings)
totalRating ← totalRating + animeRatings[1]
totalRating ← totalRating + animeRatings[2]
averageRating ← totalRating / count
DISPLAY("Running total after two shows: " + totalRating)

PROCEDURE addAnime(titleList, ratingList, statusList, title, rating, status)
{
    APPEND(titleList, title)
    APPEND(ratingList, rating)
    APPEND(statusList, status)
    RETURN(titleList)
}

PROCEDURE getLetterGrade(rating)
{
    IF (rating ≥ 9)
    {
        RETURN("S-Tier")
    }
    ELSE
    {
        IF (rating ≥ 7)
        {
            RETURN("A-Tier")
        }
        ELSE
        {
            RETURN("Needs Improvement")
        }
    }
}

animeTitles ← addAnime(animeTitles, animeRatings, animeStatus, newTitle, newRating, newStatus)
tier ← getLetterGrade(newRating)
DISPLAY(newTitle + " is rated: " + tier)

IF (newStatus = "Completed")
{
    DISPLAY(newTitle + " — nice, you finished it!")
}
ELSE
{
    IF (newStatus = "Dropped")
    {
        DISPLAY(newTitle + " — dropped. No worries, not every show is a fit.")
    }
    ELSE
    {
        DISPLAY(newTitle + " — enjoy the watch!")
    }
}

response ← "yes"
REPEAT UNTIL (response = "no")
{
    title ← INPUT("Add another anime? Enter title (or type 'done'): ")
    IF (title = "done")
    {
        response ← "no"
    }
    ELSE
    {
        rating ← INPUT("Rating (1-10): ")
        status ← INPUT("Status: ")
        APPEND(animeTitles, title)
        APPEND(animeRatings, rating)
        APPEND(animeStatus, status)
    }
}

PROCEDURE calculateAverageRating(ratingList)
{
    sum ← 0
    i ← 1
    REPEAT UNTIL (i > LENGTH(ratingList))
    {
        sum ← sum + ratingList[i]
        i ← i + 1
    }
    average ← sum / LENGTH(ratingList)
    RETURN(average)
}

avg ← calculateAverageRating(animeRatings)
IF (avg ≥ 8)
{
    DISPLAY("Great taste! Average rating: " + avg)
}
ELSE
{
    DISPLAY("Average rating: " + avg)
}

APPEND(animeTitles, "Demon Slayer")

INSERT(animeTitles, 1, "Fullmetal Alchemist: Brotherhood")

REMOVE(animeTitles, 3)

DISPLAY("Total anime tracked: " + LENGTH(animeTitles))

PROCEDURE findAnime(titleList, searchTitle)
{
    FOR EACH index IN NUMBERS(1, LENGTH(titleList))
    {
        IF (titleList[index] = searchTitle)
        {
            RETURN(index)
        }
    }
    RETURN(-1)
}

searchFor ← INPUT("Search for an anime title: ")
position ← findAnime(animeTitles, searchFor)

IF (position = -1)
{
    DISPLAY(searchFor + " is not in your tracker.")
}
ELSE
{
    DISPLAY(searchFor + " found at position " + position)
    DISPLAY("Your rating: " + animeRatings[position])
}

PROCEDURE isMustWatch(rating, status)
{
    IF ((rating ≥ 9) AND (status = "Completed"))
    {
        RETURN(true)
    }
    ELSE
    {
        RETURN(false)
    }
}

FOR EACH index IN NUMBERS(1, LENGTH(animeTitles))
{
    IF (isMustWatch(animeRatings[index], animeStatus[index]) = true)
    {
        DISPLAY(animeTitles[index] + " is a MUST-WATCH!")
    }
}

FOR EACH index IN NUMBERS(1, LENGTH(animeStatus))
{
    IF (NOT (animeStatus[index] = "Completed"))
    {
        DISPLAY(animeTitles[index] + " is still in progress or dropped.")
    }
}
```
